```python
presence_dict = {}
for item, group in filtered.groupby("item_name"):
    presence_dict[item] = infer_active_days(group["created_at"], max_gap_days=90)
presence_df = pd.concat(presence_dict, axis=1).fillna(False)

presence_weekly = presence_df.resample('W').max().fillna(False).astype(bool).to_period('W')
dish_order = list(filtered.item_name.value_counts().index)
plot_boolean_time_series(presence_weekly, loc_id, before_after_details_true, dish_order, [])
plot_dish_time_series(filtered.set_index('created_at'), loc_id, before_after_details_true, top_n=70)

presence_daily = strict_bridge_fill(presence_df, limit=7).resample('D').max()
#presence_daily = (presence_weekly.astype(float).replace(0.0, np.nan)[vegetarian_dishes].resample('D').interpolate(limit=6).fillna(0).astype(int).sum(axis=1).plot())
menu = pd.read_csv(Path("scripts") / "labeling" / "dish_labels" / (loc_id + ".csv"))

vegan_dishes = menu.loc[menu["vegan"], "item_name"]
vegetarian_dishes = menu.loc[menu["vegetarian"], "item_name"]
mpbamod_dishes = menu.loc[menu["mpbamod"], "item_name"]
lamb_dishes = menu.loc[menu["lamb"] & menu["mpbamod"], "item_name"] 
chunked_beef_or_pork_dishes = menu.loc[menu["chunked_beef_or_pork"] & menu["mpbamod"], "item_name"] 
pulled_pork_dishes = menu.loc[menu["pulled_pork"] & menu["mpbamod"], "item_name"] 
beef_or_pork_burger_dishes = menu.loc[menu["beef_or_pork_burger"] & menu["mpbamod"], "item_name"] 
ground_meat_dishes = menu.loc[menu["ground_meat"] & menu["mpbamod"], "item_name"] 
meatballs_dishes = menu.loc[menu["meatballs"] & menu["mpbamod"], "item_name"] 
sausage_dishes = menu.loc[menu["sausage"] & menu["mpbamod"], "item_name"] 
bacon_dishes = menu.loc[menu["bacon"] & menu["mpbamod"], "item_name"] 
breakfast_sausage_patty_dishes = menu.loc[menu["breakfast_sausage_patty"] & menu["mpbamod"], "item_name"] 
unfried_chicken_dishes = menu.loc[menu["unfried_chicken"] & menu["mpbamod"], "item_name"] 
fried_chicken_dishes = menu.loc[menu["fried_chicken"] & menu["mpbamod"], "item_name"] 
savory_dairy_dishes = menu.loc[menu["savory_dairy"] & menu["mpbamod"], "item_name"] 
sweet_dairy_dishes = menu.loc[menu["sweet_dairy"] & menu["mpbamod"], "item_name"] 
egg_dishes = menu.loc[menu["egg"] & menu["mpbamod"], "item_name"]

promo_date = before_after_details_true.loc[loc_id, 'cross_over_date']
indicator = pd.to_datetime(promo_date) < presence_daily.index

dishes_count = presence_daily.sum(axis=1)
vegan_dishes_count = presence_daily[vegan_dishes].sum(axis=1)
vegetarian_dishes_count = presence_daily[vegetarian_dishes].sum(axis=1)
lamb_dishes_count = presence_daily[lamb_dishes].sum(axis=1)
chunked_beef_or_pork_dishes_count = presence_daily[chunked_beef_or_pork_dishes].sum(axis=1)
pulled_pork_dishes_count = presence_daily[pulled_pork_dishes].sum(axis=1)
beef_or_pork_burger_dishes_count = presence_daily[beef_or_pork_burger_dishes].sum(axis=1)
ground_meat_dishes_count = presence_daily[ground_meat_dishes].sum(axis=1)
meatballs_dishes_count = presence_daily[meatballs_dishes].sum(axis=1)
sausage_dishes_count = presence_daily[sausage_dishes].sum(axis=1) * indicator
bacon_dishes_count = presence_daily[bacon_dishes].sum(axis=1) * indicator
breakfast_sausage_patty_dishes_count = presence_daily[breakfast_sausage_patty_dishes].sum(axis=1) * indicator
unfried_chicken_dishes_count = presence_daily[unfried_chicken_dishes].sum(axis=1)
fried_chicken_dishes_count = presence_daily[fried_chicken_dishes].sum(axis=1)
savory_dairy_dishes_count = presence_daily[savory_dairy_dishes].sum(axis=1)
sweet_dairy_dishes_count = presence_daily[sweet_dairy_dishes].sum(axis=1)
egg_dishes_count = presence_daily[egg_dishes].sum(axis=1)

mpbamod_dishes_count = presence_daily[mpbamod_dishes].sum(axis=1) * indicator

menu_counts = pd.concat([
    dishes_count, vegan_dishes_count, vegetarian_dishes_count, mpbamod_dishes_count,
    lamb_dishes_count,chunked_beef_or_pork_dishes_count,pulled_pork_dishes_count,
    beef_or_pork_burger_dishes_count,ground_meat_dishes_count,meatballs_dishes_count,
    sausage_dishes_count,bacon_dishes_count,breakfast_sausage_patty_dishes_count,
    unfried_chicken_dishes_count,fried_chicken_dishes_count,
    savory_dairy_dishes_count,sweet_dairy_dishes_count,
    egg_dishes_count
], axis=1).set_axis([
    'dishes_count', 'vegan_dishes_count', 'vegetarian_dishes_count', 'mpbamod_dishes_count', 
    'lamb_dishes_count', 'chunked_beef_or_pork_dishes_count', 'pulled_pork_dishes_count', 
    'beef_or_pork_burger_dishes_count', 'ground_meat_dishes_count', 'meatballs_dishes_count', 
    'sausage_dishes_count', 'bacon_dishes_count', 'breakfast_sausage_patty_dishes_count', 
    'unfried_chicken_dishes_count', 'fried_chicken_dishes_count', 
    'savory_dairy_dishes_count', 'sweet_dairy_dishes_count', 
    'egg_dishes_count'
    ], axis=1)
menu_counts.plot()
menu_counts.to_csv(Path("scripts") / "labeling" / "dish_counts" / (loc_id + ".csv"))
```
