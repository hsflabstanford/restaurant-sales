```python
# plot histograms of unit price of chicken and organic chicken on same plot
import matplotlib.pyplot as plt

plt.hist(data[location_ids_by_coverage[2]].query('item_name.str.contains("Chicken")')['unit_price'], alpha=0.5, label='Chicken')
plt.hist(data[location_ids_by_coverage[2]].query('item_name.str.contains("Organic Chicken")')['unit_price'], alpha=0.5, label='Organic Chicken')
plt.legend()
plt.xlabel('Unit Price')
plt.ylabel('Frequency')
plt.title('Histogram of Unit Price for Chicken and Organic Chicken')
plt.show()

# plot distribution of dates
plt.hist(data[location_ids_by_coverage[2]].query('item_name.str.contains("Chicken")')['created_at'], alpha=0.5, bins=100, label='Chicken')
plt.hist(data[location_ids_by_coverage[2]].query('item_name.str.contains("Organic Chicken")')['created_at'], alpha=0.5, bins=100, label='Organic Chicken')
plt.legend()
plt.xlabel('Order Date')
plt.ylabel('Frequency')
plt.title('Histogram of Order Date for Chicken and Organic Chicken')
plt.show()

# condition both organic chicken and chicken to be berfore the date chicken stopped being a product and compare histogram of those prices before that date
chicken_last_date = data[location_ids_by_coverage[2]].query('item_name == "Chicken"')['created_at'].max()
conditioned_data = data[location_ids_by_coverage[2]].loc[data[location_ids_by_coverage[2]]['created_at'] < chicken_last_date]
plt.hist(conditioned_data.query('item_name.str.contains("Chicken")')['unit_price'], alpha=0.5, label='Chicken')
plt.hist(conditioned_data.query('item_name.str.contains("Organic Chicken")')['unit_price'], alpha=0.5, label='Organic Chicken')
plt.legend()
plt.xlabel('Unit Price')
plt.ylabel('Frequency')
plt.title('Histogram of Unit Price for Chicken and Organic Chicken')
plt.show()
```
