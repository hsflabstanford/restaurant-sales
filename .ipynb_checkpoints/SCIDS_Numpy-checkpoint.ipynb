

### Broadcasting
Another important concept is `Broadcasting`, distributing calculations across arrays that can be of different sizes. 

`Broadcasting` follows a strict set of rules to determine the interaction between the two arrays:


```python
a = np.array([0, 1, 2])
b = np.array([5, 5, 5]) 
a + b, a + 5 # they return the same output 
```

```python
M = np.ones((4, 3))
M, a  #, M + a, M.shape, a.shape
```

```python
M + a #  M.shape, a.shape
```

```python
a = np.arange(3)
b = np.arange(3)[:, np.newaxis]

print(f'a: {a} \na.shape: {a.shape}')
print(f'b: {b} \nb.shape: {b.shape}')
a + b 
```

`Broadcasting` Rules!

- Rule 1: If the two arrays differ in their number of dimensions, the shape of the one with fewer dimensions is padded with ones on its leading (left) side.
- Rule 2: If the shape of the two arrays does not match in any dimension, the array with shape equal to 1 in that dimension is stretched to match the other shape.
- Rule 3: If in any dimension the sizes disagree and neither is equal to 1, an error is raised.

Numpy tile: [link](https://numpy.org/doc/stable/reference/generated/numpy.tile.html)


```python
a = np.arange(12).reshape(2,2,3)
b = np.arange(6).reshape(2,3)

# (Rule 1)
# print(b)
# print(np.tile(b, (2,1,1)))

a + np.tile(b, (2,1,1)) == a+b # operations are commutative, meaning that a+b=b+a.
a+b
```

```python
a
```

```python
b
```

```python
a = np.arange(24).reshape(2,2,2,3)
b = np.arange(12).reshape(2,2,3)
a+b # (Rule 1)
```

```python
a = np.arange(12).reshape(3,2,2)
b = np.arange(6).reshape(3,2,1)
print(np.tile(b, (1,1,2)))
a + np.tile(b, (1,1,2)) == a + b # (Rule 2)
```

```python
a = np.arange(12).reshape(2,2,3)
b = np.arange(6).reshape(2,1,3)

a + np.tile(b, (1,2,1)) == a + b # (Rule 2)
```

```python
a = np.arange(12).reshape(2,2,3)
b = np.arange(24).reshape(2,4,3)
# a + b # (Rule 3)

a = np.arange(12).reshape(2,2,3)
b = np.arange(4).reshape(2,2)
# a + b # (Rule 1 -> Rule 3)
```

Remark
- `Broadcasting` is often orders of magnitude faster than running loops. A common usecase is to calculate pairwise distances between two arrays.

- `Broadcasting` is frequently used in data normalization. (e.g., the original data is a 100 X 5 matrix, and its mean vector is a 5-dimensional vector).

## [Array-Functions 2] Indexing and subsetting 

Indexing in Numpy is pretty similar to the Python `list`'s indexing. Moreover, similar to Python list, we can retrieve values with the square brackets `[]`.

There are mainly three ways of accessing elements in numpy array. 
  - Access by an index counting from zero. (e.g., `arr[4]`)
  - Slice a segment of values using `:`. (e.g., `arr[:4]`)
  - Boolean mask (e.g., `arr[arr>4]`)
  

### Indexing

```python
demo1 = np.arange(12).reshape(3,4)
print(demo1)
print(f'Shape of demo1 : {demo1.shape}')
print(f'Type of demo1 : {type(demo1)}')
```

```python
# Accessing by an index
demo1[2] # third row 
demo1[-1] # index from the end of the array
demo1[-1], demo1[2] == demo1[-1]
```

```python
demo1[2, 1] # at third row and second column 
demo1[-1, 1] 
demo1[2, 1], demo1[-1, 1]  
```

```python
# Accessing by a list
demo1[[0,2]] # first and third row

# demo1[2] vs demo1[[2]]
demo1[2], demo1[[2]]
```

```python
print('Before modification: \n', demo1)
demo1[2, 1] = 1000
print('After modification: \n', demo1)
```

### Slicing

Again, NumPy slicing syntax follows that of the standard Python list with `:` character.

Slicing is used as follows. `numpy_array[start:stop:step]`


```python
# slicing
demo1 = np.arange(12)
print(demo1[:5]) # first five elements
print(demo1[5:]) # element after index 5
print(demo1[:]) # every element

print(demo1[2:6]) # middle array
print(demo1[::2]) # every other element (odd index)
print(demo1[1::2]) # every even index element
```

```python
demo1 = np.arange(12).reshape(3,4)
print(demo1[2,:2])
print(demo1[:4,:2]) # demo1[:,:2] 
print(demo1[:,:2])
```

If we subset with boolean arrays, the boolean array provided for the row must have the same size as the number of rows. The columns work similarly.

```python
# boolean mask

# Creates a boolean array
small_col1 = demo1[:, 0] < 2
print(small_col1)

# Subset the rows using a boolean array
print(demo1[small_col1]) # return the first row
```

## [NP-Functions 1] Numpy has a lot of mathematical functions built-in to it....

- `np.hstack()` or `np.vstack()` stack the the lists of equal length into an array.
- `np.concatenate()` takes as input a sequence and stacks along the axis [link](https://numpy.org/doc/stable/reference/generated/numpy.concatenate.html).
- `np.isnan` and `np.isinf` are useful to detect if there are `np.nan` or `np.inf` values.
- `np.logical_not()` negates a boolean array, `np.logical_not(np.array([True, False]))` 
- `np.where()` returns the indices that correspond to True in a boolean array. `np.where(np.array([1, 2, 3]) > 2)` 
- `np.loadtxt()` loads data from a text file (txt,csv, json, ...) from your computer [link](https://numpy.org/doc/stable/reference/generated/numpy.loadtxt.html)

```python
demo_mat=np.arange(12).reshape(3,4)
random_mat=np.random.random(12).reshape(3,4)
np.hstack((demo_mat, random_mat)), np.vstack((demo_mat, random_mat))
```

```python
np.concatenate((demo_mat, random_mat)) == np.vstack((demo_mat, random_mat))
```

```python
np.concatenate((demo_mat, random_mat), axis=1) == np.hstack((demo_mat, random_mat))
```

```python
# np.apply_along_axis
col_sum = np.apply_along_axis(np.sum, 0, demo_mat)
print(col_sum)
```

```python
negation_vector=np.logical_not(np.array([True, False, False]))
negation_vector[np.where(negation_vector)]
```

```python
negation_mat=(np.random.normal(size=(3,4,5)) > 0).astype(bool)
negation_mat[np.where(negation_mat)]
```

## [NP-Functions 2] Advanced linear algebra functions

- `np.linalg.inv` : Matrix inversion. 
- `np.linalg.eig` : Eigenvalues and eigenvectors
- `np.linalg.svd` : Singular value decomposition

```python
gaussian_matrix=np.random.normal(size=(3,100))
gram_matrix=gaussian_matrix.dot(gaussian_matrix.T) 
gram_matrix
```

```python
np.linalg.inv(gram_matrix)
```

```python
eigenvalues, eigenvectors = np.linalg.eig(gram_matrix) # two outputs
eigenvalues
```

```python
np.linalg.svd(gram_matrix) # three outputs
```
