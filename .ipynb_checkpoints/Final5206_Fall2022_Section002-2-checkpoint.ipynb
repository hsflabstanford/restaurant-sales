Our goal is to wrangle the data into a data frame where the rows correspond to different companies and the columns are:
- `name`: The company name as in `employer2stock`, these should be strings
- `symbol`: The stock symbol as in `employer2stock`, these should be strings
- `adj_close`: The monthly adjusted close price, these should be floats
- `date`: The end date of the adjusted close price in that month, these should be datetime objects
- `year`: The year for that corresponding record, these should be integers

For grading:
- store the final output as a variable called `stocks_df`
- print out the number of rows in a human readable message
- print out the first 3 rows in `stocks_df`
- print out the data types for each column in `stocks_df`

````python
employer2stock = {
    "INFOSYS LIMITED": "INFY",
    "ACCENTURE LLP": "ACN",
    "NTT DATA": "NTDTY",
    "MICROSOFT": "MSFT",
    "AMAZON WEB SERVICES": "AMZN",
    "INTEL": "INTC",
    "NVIDIA": "NVDA",
    "ORACLE AMERICA": "ORCL",
    "QUALCOMM TECHNOLOGIES": "QCOM",
    "VMWARE": "VMW"
}
````

````python
ts_data = json.load(open("stocks_monthly_adjusted.json", "r"))
````

````python
ts_data.keys()
````

````python
list_of_dfs = []
for name in employer2stock:
    
    symbol = employer2stock[name]
    
    adj_close = pd.DataFrame(ts_data[symbol]['Monthly Adjusted Time Series']).loc["5. adjusted close"].reset_index()
    adj_close.columns = ['date','adj_close']
    adj_close['date'] = adj_close['date'].apply(lambda s: pd.to_datetime(s))
    adj_close['year'] = adj_close['date'].dt.year
    
    n_obs = adj_close.shape[0]
    
    name_symbol = pd.DataFrame(columns = ['name','symbol'], data = np.array([[name]*n_obs,[symbol]*n_obs]).T)
    
    df = pd.concat([name_symbol, adj_close], axis = 1)
    
    list_of_dfs.append(df)

stocks_df = pd.concat(list_of_dfs, axis = 0)
````

````python
stocks_df['adj_close'] = stocks_df['adj_close'].astype('float')
````

````python
print(f'Number of rows in stocks_df: {stocks_df.shape[0]}')
````

## Q7 - Modeling trends (20 points)

If you couldn't finish Q5, please use `Q5_backup_final_fall2022.csv` for `jdf`.

- For each employer in `jdf`, please fit 2 models for `Initial Approval` using oridinary least squares regression ([sklearn.linear_model.LinearRegression](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LinearRegression.html)) but limited to data in 2020 and earlier. The models should be:
  - Model 1 should regress on `year`, `avg_adj_close`, and `std_adj_close`
  - Model 2 should regress on `year` alone.
  Please ignore the issue of forecasting for the sake of the exam.
- Please show which model predicts the outcomes in 2021 better with a reasonable metric and a human-readable message.
- From the winning model, please store the coefficient corresponding to `Fiscal Year` (we'll call this the "trend") and the prediction at year 2021. Store your results in a data frame called `output` with columns: `name`, `trend`, `pred_2021`.
  - For clarification, `pred_2021` is the linear model predictions of `Initial Approval` in 2021

````python
#jdf = pd.read_csv("Q5_backup_final_fall2022.csv")
````

````python
jdf.tail()
````

````python
before_2021 = jdf[jdf['year'] < 2021]
````

````python
data_2021 = jdf[jdf['year'] == 2021]
````

````python
data_2021.head()
````

Let's pick rmse as our metric, and see which model is chosen with respect to this metric more

````python
def rmse(actual, pred):
    return (((actual - pred)**2).sum())**(1/2)
````

````python
employers = before_2021['name'].tolist()
trend = []

model1_count = 0
model2_count = 0
for employer in employers:
    
    employer_data_before = before_2021[before_2021['name'] == employer]
    employer_data_2021 = data_2021[data_2021['name'] == employer]
    
    y_train = employer_data_before.loc[:,['Initial Approval']].values
    y_test = employer_data_2021.loc[:,['Initial Approval']].values
    
    X1_train = employer_data_before.loc[:,['year', 'avg_adj_close', 'std_adj_close']].values
    X1_test = employer_data_2021.loc[:,['year', 'avg_adj_close', 'std_adj_close']].values
    
    X2_train = employer_data_before.loc[:,['year']].values
    X2_test = employer_data_2021.loc[:,['year']].values
    
    lr1 = LinearRegression()
    lr1.fit(X1_train, y_train)
    lr2 = LinearRegression()
    lr2.fit(X2_train, y_train)
    
    pred1 = lr1.predict(X1_test)
    pred2 = lr2.predict(X2_test)
    
    #print(X1_test)
    #print(pred1)
    
    rmse1 = rmse(y_test, pred1)
    rmse2 = rmse(y_test, pred2)
    
    winner = 0
    
    if rmse1 < rmse2:
        winner = 1
    else:
        winner = 2
    
    coef = 0
    
    if winner == 1:
        coef = lr1.coef_[0,0]
        pred = pred1[0,0]
        model1_count += 1
    else:
        coef = lr2.coef_[0,0]
        pred = pred2[0,0]
        model2_count += 1
        
        
    trend.append((employer,coef,pred))     
````

````python
model1_win_pct = round(100*(model1_count/len(employers)))
model2_win_pct = round(100*(model2_count/len(employers)))
````

````python
# With respect to rmse
print(f'Model 1 was better {model1_win_pct}% of the time')
print(f'Model 2 was better {model2_win_pct}% of the time')
````

````python
output = pd.DataFrame(columns = ['name', 'trend', 'pred_2021'], data = trend)
````
