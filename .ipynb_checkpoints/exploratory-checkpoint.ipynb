Sales data item name and unit price (non)uniqueness

```python
def find_non_identical_duplicates_by_id(df, location_id):
    # Count the number of unique values in each column per group
    nunique_per_group = df.groupby('unique_id').nunique()

    # Identify unique_ids with more than one unique value in any column
    non_unique_ids = nunique_per_group.max(axis=1) > 1
    non_unique_ids = non_unique_ids[non_unique_ids].index

    # Exclude groups where all rows are identical
    total_rows_per_group = df.groupby('unique_id').size()
    non_identical_groups = total_rows_per_group[non_unique_ids] != nunique_per_group.loc[non_unique_ids].max(axis=1)
    filtered_unique_ids = non_identical_groups[non_identical_groups].index

    # Filter the data to include only those unique_ids
    non_unique_data = df[df['unique_id'].isin(filtered_unique_ids)]

    # Count exact duplicates in each group
    exact_duplicates_count = df[df.duplicated(keep=False)].groupby('unique_id').size()

    # Prepare a summary of non-unique items
    summary = {
        'location_id': location_id,
        'non_unique_items': len(filtered_unique_ids),
        'non_unique_amount': non_unique_data.shape[0],
        'exact_duplicates': exact_duplicates_count.sum()  # Total count of exact duplicates
    }
    return non_unique_data, summary
```

```python
def find_non_identical_duplicates_by_unit_price(data, location_id):
    
    # Find near-duplicates and remove
    data.sort_values(['item_name', 'unit_price'], inplace=True)
    data['price_diff'] = data.groupby('item_name')['unit_price'].diff().abs()
    data = data[(data['price_diff'] >= 2) | (data['price_diff'].isna())] # Treat differences of less than 

    # Determine which items have multiple prices (with price differences greater than 2 cents) and filter to only those
    unique_price_counts = data.groupby('item_name')['unit_price'].nunique()
    items_with_multiple_prices = unique_price_counts[unique_price_counts > 1].index
    data = data[data['item_name'].isin(items_with_multiple_prices) & ~data['item_name'].str.contains('nan')]
    non_unique = data.drop_duplicates(subset=['item_name', 'unit_price'])

    return non_unique
```

```python
# Applying the function to each DataFrame in the dataset
non_identical_duplicates_list = []
summary_list = []
# By unit price
non_unique_list = []
non_unique_amount_list = []
for location_id, df in tqdm(restaurant_sales_data.items()):
    
    # Copy data to avoid overwriting
    data = df.copy()
    non_unique_data, summary = find_non_identical_duplicates_by_id(data, location_id)
    non_identical_duplicates_list.append(non_unique_data)
    summary_list.append(summary)

    non_unique = find_non_identical_duplicates_by_unit_price(data, location_id)
    non_unique_list.append(non_unique)
    non_unique_amount_list.append({'location_id' : location_id, 'non_unique_items' : non_unique['item_name'].nunique(), 'non_unique_amount' : non_unique.shape[0]})

# Creating a summary DataFrame
non_identical_duplicates_summary = pd.DataFrame(summary_list)
non_unique_amounts = pd.DataFrame(non_unique_amount_list)
```

```python
# Write results to an Excel file
#if not os.path.exists('example_palate_data_issues/non_unique_items.xlsx'):
with pd.ExcelWriter('example_palate_data_issues/non_unique_items.xlsx') as writer:
    non_identical_duplicates_summary.to_excel(writer, sheet_name='duplicate_by_id_amounts')
    non_unique_amounts.to_excel(writer, sheet_name='non_unique_amounts')
    for location_id, df_original in tqdm(zip(location_ids, non_unique_list)):
        df = df_original.copy()
        df.index = df.index.astype(str)
        df.to_excel(writer, sheet_name=location_id)
```

### Merging Data

Merging the sales data and the menu data

```python
# Check if 'items_tagged' has unique pairs of 'item_name' and 'location_id'
if items_tagged.duplicated(subset=['item_name', 'location_id']).any():
    raise ValueError("Duplicates found in 'items_tagged' for the combination of 'item_name' and 'location_id'")

merged_sales_and_menu = {}
for location_id, df in tqdm(restaurant_sales_data.items()):

    # Copy to prevent overwriting
    data = df.copy()
    items_tagged_copy = items_tagged.copy()
    
    # Perform the merge on both 'item_name' and 'location_id'
    merged = pd.merge(data.reset_index(), items_tagged_copy, 
                      on=['item_name', 'location_id'], how='left')
    
    # Set 'created_at' back as the index
    merged.set_index('created_at', inplace=True)

    # Store in the dictionary
    merged_sales_and_menu[location_id] = merged
```

```python
num_plots = len(merged_sales_and_menu)
cols = 4 
rows = math.ceil(num_plots / cols) * 4

# Create a figure with multiple subplots
fig, axs = plt.subplots(rows, cols, figsize=(15, 5 * rows))
axs = axs.flatten()  # Flatten the array for easy indexing

top_dishes_number = 20

for i, (location_id, data) in enumerate(merged_sales_and_menu.items()):

    df = data.copy()
    
    # Get the top 20 items sorted in descending order
    top_items_by_times_ordered = df['item_name'].value_counts().nlargest(top_dishes_number).sort_values()

    # Plot in the appropriate subplot as a horizontal bar chart
    axs[4*i].barh(top_items_by_times_ordered.index.str.slice(0,20).str.capitalize(), top_items_by_times_ordered)
    axs[4*i].set_title(f'{location_id}\nTotal by Times Ordered')
    
    # Set y-axis label and make y-labels smaller
    axs[4*i].tick_params(axis='y', labelsize=10)
    axs[4*i].set_xlabel('Times Ordered')

    top_items_by_quantity_ordered = df.groupby('item_name')['item_quantity'].sum().nlargest(top_dishes_number).sort_values()
    
    # Plot in the appropriate subplot as a horizontal bar chart
    axs[4*i + 1].barh(top_items_by_quantity_ordered.index.str.slice(0,20).str.capitalize(), top_items_by_quantity_ordered, color='cyan')
    axs[4*i + 1].set_title(f'{location_id}\nTotal by Quantity Ordered')
    
    # Set y-axis label and make y-labels smaller
    axs[4*i + 1].tick_params(axis='y', labelsize=10)
    axs[4*i + 1].set_xlabel('Quantity Ordered')

    plant_df = df[df['is_plant_based'] == 'yes']
    
    # Get the top 20 items sorted in descending order
    top_plant_based_items_by_times_ordered = plant_df['item_name'].value_counts().nlargest(top_dishes_number).sort_values()

    # Plot in the appropriate subplot as a horizontal bar chart
    axs[4*i + 2].barh(top_plant_based_items_by_times_ordered.index.str.slice(0,20).str.capitalize(), top_plant_based_items_by_times_ordered, color='green')
    axs[4*i + 2].set_title(f'{location_id}\nPlant-Based by Times Ordered')
    
    # Set y-axis label and make y-labels smaller
    axs[4*i + 2].tick_params(axis='y', labelsize=10)
    axs[4*i + 2].set_xlabel('Times Ordered')

    top_plant_based_items_by_quantity_ordered = plant_df.groupby('item_name')['item_quantity'].sum().nlargest(top_dishes_number).sort_values()
    
    # Plot in the appropriate subplot as a horizontal bar chart
    axs[4*i + 3].barh(top_plant_based_items_by_quantity_ordered.index.str.slice(0,20).str.capitalize(), top_plant_based_items_by_quantity_ordered, color='lime')
    axs[4*i + 3].set_title(f'{location_id}\nPlant-Based by Quantity Ordered')
    
    # Set y-axis label and make y-labels smaller
    axs[4*i + 3].tick_params(axis='y', labelsize=10)
    axs[4*i + 3].set_xlabel('Quantity Ordered')

# Add a global title at the top of the figure
fig.suptitle('Distribution of Plant-Based and Total Item Orders in Each Restaurant', fontsize=16)

# Adjust layout and save the figure
plt.tight_layout()
plt.subplots_adjust(top=0.95)  # Adjust the top margin to make room for the global title
plt.savefig('Restaurant Both Sales Distributions.png', bbox_inches='tight')
```

```python
import matplotlib.pyplot as plt
import math

# Assuming merged_sales_and_plant_based and merged_sales_and_another_criteria are defined and have the same length
num_plots_per_dict = len(merged_sales_and_plant_based)
cols = 3
rows = math.ceil(num_plots_per_dict / cols) * 2  # Total number of rows for both dictionaries

# Create a figure with multiple subplots
fig, axs = plt.subplots(rows, cols, figsize=(15, 5 * rows))
axs = axs.reshape(rows, cols)  # Reshape for easier indexing

def plot_bar_chart(ax, data, title, color):
    """Plots a horizontal bar chart on the given axes."""
    top_items = data['item_name'].value_counts().nlargest(20).sort_values()
    ax.barh(top_items.index.str.slice(0, 20).str.capitalize(), top_items, color=color)
    ax.set_title(title)
    ax.tick_params(axis='y', labelsize=10)
    ax.set_xlabel('Orders')

# Iterate over the rows and plot data from each dictionary alternately
for row in range(rows):
    if row % 2 == 0:  # Even rows: merged_sales_and_plant_based
        dict_to_use = merged_sales_and_plant_based
    else:  # Odd rows: merged_sales_and_another_criteria
        dict_to_use = merged_sales_and_plant_based

    for col in range(cols):
        index = (row // 2) * cols + col
        if index < num_plots_per_dict:
            location_id, df = list(dict_to_use.items())[index]
            plot_bar_chart(axs[row, col], df, location_id, 'green' if row % 2 == 0 else 'blue')

# Add a global title at the top of the figure
fig.suptitle('Distribution of Item Orders in Each Restaurant', fontsize=16)

# Adjust layout and save the figure
plt.tight_layout()
plt.subplots_adjust(top=0.95)
plt.savefig('Restaurant Sales Distributions.png', bbox_inches='tight')
```
