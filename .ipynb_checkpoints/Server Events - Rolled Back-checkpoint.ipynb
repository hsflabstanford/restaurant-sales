## Goal

- The aim of this notebook is to understand churn over the course of Dobility's existence.
- We want to catalog each time a customer (account) starts and stops a paid subscription.

Import Server Data

```python
names = ["Name","ID","Type","Subject","Date","Status","Subscription","Account","Subscription Activation Date"]
dtypes = {'Type':'category',
          'Status':'category',
          'Subscription':'category'
}
events = pd.read_csv('Server Events.csv', header = 0, names = names, dtype = dtypes, parse_dates = ['Date','Subscription Activation Date'])
```

### Chronological Order

If the events are sorted by ID, then they become organized chronologically from 2013 to 2022.

```python
events.sort_values('ID', inplace = True)
events.reset_index(drop = True, inplace = True)
```

Additional Type Formatting

```python
events['Account'] = events['Account'].astype(str)
```

### Irrelevant Attributes

- *Status*, *Subscription*, and *Subscription Activation Date* are useless for creating a timeline of paid servers because they are all **post hoc** labels.
- That is, they only tell us the server's last state.

- For instance, the server **oxfamnicaragua**'s *Subscription* is community, yet it had a paid subscription for many months prior.
- Conversely, the *Subscription Activation Date* is the date of the first subscription activation, which may be community or paid, so it is unhelpful as well.

```python
events.drop(['Status','Subscription','Subscription Activation Date'], axis = 1, inplace = True)
```

Making an Alphabet String

```python
alphabet_list = []
# ASCII upper case codes
uppercase = list(range(65,(65+26)))   
# ASCII lower case codes
lowercase = list(range(97,(97+26)))
# ASCII conversion
for i in (uppercase + lowercase):
    alphabet_list.append(chr(i))
# list -> string
alphabet = ''.join(alphabet_list)   
```

```python
class AccountMap:
    
    def __init__(self, data: pd.DataFrame):
        
        self.data = data   # input df
        self.map = {}
        self.total_paid_accounts = 0
        self.total_comm_accounts = 0
        self.acc_df = pd.DataFrame()
        self.team_df = pd.DataFrame()
        self.server_dfs = {}   # output dfs
        
    def set_map(self, filled_map: dict):
        
        self.map = filled_map
        
    def get_map(self):
        
        return self.map
    
    # creates a dictionary of paid accounts with their respective time series of servers and teams
    # key = account
    # value = list of tuples for each date, where each tuple is:
        # 1. date of event (DoE) , 
        # 2. # of teams in all servers on DoE, and 
        # 3. dictionary of server-team pairs
    # ex. current_accounts['Oxfam'] = (
        # 2015-11-15 00:52:00, 
        # 4, 
        # {('oxfamsom',2),('oxfamght',1),('oxfamsicily',1)})

    def create_paid(self):

        self.map['total!'] = [(pd.to_datetime(self.data['Date'].iloc[0]),self.total_paid_accounts)]

        for i in range(self.data.shape[0]):

            row = self.data.iloc[i,:]   # select the ith event row

            event = row['Type']
            account = row['Account']
            subject = row['Subject']
            date = row['Date']

            # event 1: paid activation
            paid = event == 'SUBSCRIBED' and 'Activated a' in subject
            pro = event == 'SUBSCRIBED' and '_PRO' in subject
            plus = event == 'SUBSCRIBED' and '_PLUS' in subject
            onboarded = paid or pro or plus
            
            # event 2: server closure
            closed = event == 'ACCTCLOSED' or event == 'CLOSEACCT'
            downgraded = event == 'SUBSCRIBE' and ('INTRO_COMMUNITY' in subject or 'INTRO ' in subject or 'INTRO_TRIAL' in subject)
            churned =  closed or downgraded

            if account not in self.map and onboarded:
                    
                    self.map[account] = []

            if account in self.map:
                     
                if onboarded:
                    self.add_paid(row)

                if churned:
                    self.drop_paid(row)


    # adds paid accounts (and their servers and teams) to our account dictionary
    # does so when there are SUBSCRIBED events that have a paid activation description
    # e.g. 'Activated a single-team subscription, including 1 team-month.'

    def add_paid(self, row: pd.Series):

        account = row['Account']
        server = row['Name']
        subject = row['Subject']
        date = row['Date']
        teams = 1
    
        # subscriptions with more than 1 team/month
        if 'Activated a' in subject:
            team_string = subject[10:15].strip(alphabet + ' -')   # remove alphabetic characters
            if team_string:
                teams = int(team_string)

        time_series = self.map[account]

        # select the last time step
        if time_series:
            total_teams = time_series[-1][1]
            servers = time_series[-1][2].copy()
        else:
            total_teams = 0
            servers = {}

        previous_total_teams = total_teams
        previous_total_accounts = self.total_paid_accounts
            
        if server not in servers :
            total_teams = total_teams + teams
            servers[server] = teams

        if servers[server] != teams:
            total_teams = total_teams - servers[server] + teams
            servers[server] = teams
        
        if 0 < total_teams and previous_total_teams == 0:
            self.total_paid_accounts += 1
        
        # updates specificed account's server and team time series if there was a change
        if total_teams != previous_total_teams:
            self.map[account] += [(date, total_teams, servers)]   # appends time step with new or lost teams and servers
            
        # updates total account time series if there was a change
        if self.total_paid_accounts != previous_total_accounts:
            self.map['total!'] += [(date, self.total_paid_accounts)]   # appends time step with a new or lost account
            
            
    # removes paid accounts (and their servers and teams) to our account dictionary
    # does so when all servers have been removed
    # and removes servers when there is a closure or a downgrade
    # e.g. 'User <server> a subscribed to plan INTRO_COMMUNITY'

    def drop_paid(self, row: pd.Series):

        account = row['Account']
        server = row['Name']
        date = row['Date']

        # select the last time step, discard the old date
        time_series = self.map[account]
        total_teams  = time_series[-1][1]
        servers = time_series[-1][2].copy()     

        previous_total_teams = total_teams
        previous_total_accounts = self.total_paid_accounts
        
        if server in servers:
            total_teams = total_teams - servers[server]
            servers.pop(server, None)
        
        if total_teams == 0 and 0 < previous_total_teams:
            self.total_paid_accounts -= 1
        
        # updates specificed account's server and team time series if there was a change
        if total_teams != previous_total_teams:
            self.map[account] += [(date, total_teams, servers)]   # appends time step with new or lost teams and servers
        
        # updates total account time series if there was a change
        if self.total_paid_accounts != previous_total_accounts:
            self.map['total!'] += [(date, self.total_paid_accounts)]   # appends time step with a new or lost account
            

    def create_comm(self):
        
        self.map['total!'] = [(pd.to_datetime(self.data['Date'].iloc[0]),self.total_comm_accounts)]

        for i in range(self.data.shape[0]):

            row = self.data.iloc[i,:]   # select the ith event row

            event = row['Type']
            account = row['Account']
            subject = row['Subject']
            date = row['Date']
            
            # event 1: paid activation
            onboarded = event == 'SUBSCRIBED' and 'INTRO_COMMUNITY' in subject
            
            # event 2: upgrade request
            paid = event == 'SUBSCRIBED' and 'Activated a' in subject
            pro = event == 'SUBSCRIBED' and '_PRO' in subject
            plus = event == 'SUBSCRIBED' and '_PLUS' in subject
            upgraded = paid or pro or plus
            
            # event 3: server closure
            closed = event == 'ACCTCLOSED' or event == 'CLOSEACCT'
            downgraded = event == 'SUBSCRIBED' and ('INTRO ' in subject or 'INTRO_TRIAL' in subject)
            churned =  closed or downgraded
            
            # event 4: community engagement 
            engaged = event == 'ENGAGED'
            
            if account not in self.map and onboarded:
                
                self.map[account] = []
            
            if account in self.map:
            
                if onboarded:
                    self.add_comm(row)
            
                if churned:
                    self.drop_comm(row)
                
                if upgraded:
                    self.drop_comm(row)   # change to << self.do_upgrade(row) >> later
            
                if engaged:
                    self.do_engage(row)
        
    def add_comm(self, row: pd.Series):
        
        account = row['Account']
        server = row['Name']
        date = row['Date']

        account_added = False
        server_added = False
        status = 1
        
        time_series = self.map[account]
        
        # select the last time step
        if time_series:
            total_engages = time_series[-1][2]
            servers = time_series[-1][3].copy() 
        else: 
            total_engages = 0
            servers = {}
            
        if not servers:
            account_added = True
        
        if server not in servers:
            servers[server] = 0
            server_added = True
        
        if server_added:
            self.map[account] += [(date, status, total_engages, servers)]
        
        # updates total account time series if there was a change
        if account_added:
            self.total_comm_accounts += 1
            self.map['total!'] += [(date, self.total_comm_accounts)]   # appends time step with a new or lost account
    
    
    # loses tracked server-specific engagements even when you temporarily drop the server
    
    def drop_comm(self, row: pd.Series):
        
        account = row['Account']
        server = row['Name']
        date = row['Date']

        server_dropped = False
        account_dropped = False
        status = 1
        
        time_series = self.map[account]
        
        # select the last time step
        if time_series:
            total_engages = time_series[-1][2]
            servers = time_series[-1][3].copy()
            
        if server in servers:
            servers.pop(server,None)
            server_dropped = True
            
        if not servers:
            status = 0
            if server_dropped:
                account_dropped = True
            
        if server_dropped:
            self.map[account] += [(date, status, total_engages, servers)]
        
        # updates total account time series if there was a change
        if account_dropped:
            self.total_comm_accounts -= 1
            self.map['total!'] += [(date, self.total_comm_accounts)]   # appends time step with a new or lost account
        
    def do_upgrade(self, row: pd.Series):
        
        account = row['Account']
        server = row['Name']
        date = row['Date']
        
        status = 2
        
        time_series = self.map[account]

        # select the last time step
        if time_series:
            total_engages = time_series[-1][2]
            servers = time_series[-1][3].copy()
        
        servers[server] = -1
            
        self.map[account] += [(date, status, total_engages, servers)]
    
    def do_engage(self, row: pd.Series):
        
        account = row['Account']
        server = row['Name']
        date = row['Date']
        
        server_added = False
        account_added = False
        status = 1
        
        time_series = self.map[account]

        # select the last time step
        if time_series:
            total_engages = time_series[-1][2]
            servers = time_series[-1][3].copy()
        else:
            total_engages = 0
            servers = {}
        
        if not servers:
            account_added = True
        
        if server not in servers:
            servers[server] = 0
            
        if server in servers:
            servers[server] += 1
            total_engages += 1
            
        self.map[account] += [(date, status, total_engages, servers)]
        
        # updates total account time series if there was a change
        if account_added:
            self.total_comm_accounts += 1
            self.map['total!'] += [(date, self.total_comm_accounts)]   # appends time step with a new or lost account
            
    def to_dataframe(self, choice: str = 'ats'):
        
        if not self.map:
            
            print('No account map found! First create an account map.')
            return
            
        else:
            
            if not ('a' in choice or 't' in choice or 's' in choice):
                
                print('Not a valid character set! Input a string containing a, t, and/or s.')
                return
            
            first_date = pd.to_datetime(self.data['Date'].iloc[0])
            last_date = pd.to_datetime(self.data['Date'].iloc[-1])
            lifespan = last_date.toordinal() - first_date.toordinal()
            
            template_df = pd.DataFrame(data = np.zeros(lifespan + 1), 
                  columns = ['to_drop'], 
                  index = pd.date_range(start = first_date, 
                                        end = last_date, 
                                        name = 'dates',
                                        freq = 'd', 
                                        normalize = True))
            
            ###
            # Accounts
            ###
            if 'a' in choice:
            
                total_ts = self.map['total!']
                dates, total_accounts = zip(*total_ts)   # unzip

                # df
                acc_df = pd.DataFrame(data = list(total_accounts), 
                                           columns = ['Total Accounts'], 
                                           index = pd.DatetimeIndex(data = dates, name = 'dates').normalize())

                # merge
                acc_df = pd.merge(template_df, acc_df, on = 'dates', how = 'outer')

                acc_df.fillna(method = 'ffill', inplace = True)
                acc_df.fillna(0, inplace = True)
                acc_df.drop(columns = 'to_drop', inplace = True)

                self.acc_df = acc_df.copy()
            
            ###
            # Teams
            ###
            if 't' in choice:
            
                team_df = template_df.copy()

                for acc in self.map:
  
                    if acc != 'total!':
                        
                        acc_ts = self.map[acc]
                        if 'p' in choice:
                            dates, total_teams, servers = zip(*acc_ts)   # unzip
                        if 'c' in choice:
                            dates, _, total_teams, servers = zip(*acc_ts)
                        
                        # df
                        each_team_df = pd.DataFrame(data = list(total_teams),
                                              columns = [acc],
                                              index = pd.DatetimeIndex(data = dates, name = 'dates').normalize())

                        # merge
                        team_df = pd.merge(team_df, each_team_df, on = 'dates', how = 'outer')

                team_df.fillna(method = 'ffill', inplace = True)
                team_df.fillna(0, inplace = True)
                team_df.drop(columns = 'to_drop', inplace = True)

                self.team_df = team_df.copy()
            
            ###
            # Servers
            ###
            if 's' in choice:
                
                for acc in self.map:
                    
                    if acc != 'total!':
                        
                        acc_ts = self.map[acc]
                        if 'p' in choice:
                            dates, total_teams, servers = zip(*acc_ts)   # unzip
                        if 'c' in choice:
                            dates, _, total_teams, servers = zip(*acc_ts)

                        dfs_list = []
                        
                        i = 0
                        for server_dict in servers:
                            
                            date = dates[i]
                            
                            # df
                            server_row = pd.DataFrame(pd.Series(server_dict, name = date)).T
                            dfs_list.append(server_row)
                            i += 1
                            
                        # concat
                        server_df = pd.concat(dfs_list, sort=False)
                        server_df.fillna(0, inplace = True)
                        
                        server_df.index.name = 'dates'

                        # merge
                        server_df = pd.merge(template_df, server_df, on = 'dates', how = 'outer')

                        server_df.fillna(method = 'ffill', inplace = True)
                        server_df.fillna(0, inplace = True)
                        server_df.drop(columns = 'to_drop', inplace = True)
                    
                self.server_dfs[acc] = server_df.copy()
        
        return 
        
        
    def visualize(self, choice: str = 'a'):
        
        if self.acc_df.empty and self.team_df.empty and not self.server_dfs:
            
            if not self.map:
                print('Nothing to plot! First create an account map.')
                return
                
            else:
                print('Nothing to plot! First convert given account map into a dataframe.')
                return
                
        else:
            
            if not ('a' in choice or 't' in choice or 's' in choice):
                
                print('Not a valid character set! Input a string containing a, t, and/or s.')
                return
            
            frame = self.acc_df.copy()   # default
            if 't' in choice:
                frame = self.team_df.copy()   # CURRENTLY BROKEN IF DF NOT MADE
            if 's' in choice:
                frame = self.server_dfs.copy()
                
            # fig, ax = plt.subplots()
            
            for i, column in enumerate(frame.columns):
                x = frame.index
                y = frame.iloc[:,i]
                plt.plot(x, y, label = column) # ax.plot vs plt.plot
                #leg = ax.legend()
            plt.legend()
        
```
