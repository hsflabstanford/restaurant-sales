```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from mpl_toolkits.mplot3d import Axes3D
```

```python
data = pd.read_csv('Soil Data.csv')
data['XRF (ppm)'] = data['XRF (ppm)'].replace(['LOD'], 0)
data_essential = data.drop(['Depth','Sampled','Analyzed'], axis = 1)
data_no_missing = data_essential.dropna()
data_numbers = data_no_missing.astype(np.float)
```

```python
data_numbers.dtypes
data_numbers.head()
```

```python
data_one_removed = data_numbers.drop(data_numbers['XRF (ppm)'].idxmax(), axis = 0).reset_index()
data_two_removed = data_one_removed.drop(data_one_removed['XRF (ppm)'].idxmax(), axis = 0).reset_index()
```

```python
X, y = data_numbers.drop(['Soil (ppm)'], axis = 1).values, data_numbers['Soil (ppm)'].values
X_one_removed, y_one_removed = data_one_removed.drop(['Soil (ppm)'], axis = 1).values, data_one_removed['Soil (ppm)'].values
X_two_removed, y_two_removed = data_two_removed.drop(['Soil (ppm)'], axis = 1).values, data_two_removed['Soil (ppm)'].values
```

```python
sns.set_style("white")

fig = plt.figure(figsize=(17, 17))
ax1 = fig.add_subplot(321, projection='3d')
ax2 = fig.add_subplot(322, projection='3d')
ax3 = fig.add_subplot(323, projection='3d')
ax4 = fig.add_subplot(324, projection='3d')
ax5 = fig.add_subplot(325, projection='3d')
ax6 = fig.add_subplot(326, projection='3d')

mean = 3
number = 50

x1 = data_numbers['X Loc'].values
y1 = data_numbers['Y Loc'].values
z1 = data_numbers['Soil (ppm)'].values
z2 = data_numbers['XRF (ppm)'].values

ax1.scatter3D(x1, y1, z1, s = 100)
ax2.scatter3D(x1, y1, z2, s = 100, c = 'orange')

ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('Soil (ppm)')
ax1.set_title('Soil', fontdict = {'fontsize': 40, 'fontweight' : 20})

ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('XRF (ppm)')
ax2.set_title('XRF', fontdict = {'fontsize': 40, 'fontweight' : 20})

x1 = data_one_removed['X Loc'].values
y1 = data_one_removed['Y Loc'].values
z1 = data_one_removed['Soil (ppm)'].values
z2 = data_one_removed['XRF (ppm)'].values

ax3.scatter3D(x1, y1, z1, s = 100)
ax4.scatter3D(x1, y1, z2, s = 100, c = 'orange')

ax3.set_xlabel('x')
ax3.set_ylabel('y')
ax3.set_zlabel('Soil (ppm)')
ax3.set_title('Soil', fontdict = {'fontsize': 40, 'fontweight' : 20})

ax4.set_xlabel('x')
ax4.set_ylabel('y')
ax4.set_zlabel('XRF (ppm)')
ax4.set_title('XRF', fontdict = {'fontsize': 40, 'fontweight' : 20})

x1 = data_two_removed['X Loc'].values
y1 = data_two_removed['Y Loc'].values
z1 = data_two_removed['Soil (ppm)'].values
z2 = data_two_removed['XRF (ppm)'].values

ax5.scatter3D(x1, y1, z1, s = 100)
ax6.scatter3D(x1, y1, z2, s = 100, c = 'orange')

ax5.set_xlabel('x')
ax5.set_ylabel('y')
ax5.set_zlabel('Soil (ppm)')
ax5.set_title('Soil', fontdict = {'fontsize': 40, 'fontweight' : 20})

ax6.set_xlabel('x')
ax6.set_ylabel('y')
ax6.set_zlabel('XRF (ppm)')
ax6.set_title('XRF', fontdict = {'fontsize': 40, 'fontweight' : 20})

plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
slr = LinearRegression()
slr.fit(X[:,[0]],y)
yhat = slr.predict(X[:,[0]])
e = y - yhat

slr1 = LinearRegression()
slr1.fit(X_one_removed[:,[0]], y_one_removed)
yhat1 = slr1.predict(X_one_removed[:,[0]])
e1 = y_one_removed - yhat1

slr2 = LinearRegression()
slr2.fit(X_two_removed[:,[0]], y_two_removed)
yhat2 = slr2.predict(X_two_removed[:,[0]])
e2 = y_two_removed - yhat2
```

```python
fig = plt.figure(figsize=(16, 3))
ax1 = fig.add_subplot(1,3,1)
ax2 = fig.add_subplot(1,3,2)
ax3 = fig.add_subplot(1,3,3)
ax1.plot(yhat , e, 'o')
ax2.plot(yhat1 , e1, 'o')
ax3.plot(yhat2, e2, 'o')

plt.show()
```
