YOUR SOLUTIONS HERE

$||Av_i||^2=(Av_i)(Av_i)=(Av_i)^\top(Av_i)=v_i^\top A^\top Av_i$

All eigenvalues of $A^\top A$ are guaranteed to be nonnegative. Let $\mathbf v_i$ be a unit eigenvector of $A^\top A$ corresponding to $\lambda_i$. Then:

$v_i^\top A^\top Av_i=v_i^\top(\lambda_iv_i)=\lambda_i||v_i||^2$

This implies that $\lambda_i$ must be negative, or $AA^\top$ guaranteed to have nonnegative eigenvalues.

2.

$AA^\top \mathbf{u}_i = \lambda_i \mathbf{v}_i$.

Start by manipulating the equation $A^\top \mathbf{u}_i = \sigma_i \mathbf{v}_i$. 

Multiply both side by $A$:

$A (A^\top \mathbf{u}_i) = A (\sigma_i \mathbf{v}_i)$


$(A A^\top) \mathbf{u}_i = \sigma_i (A \mathbf{v}_i)$

Since $\lambda_i$ and $\mathbf{u}_i$ are an eigenvalue and eigenvector of $A A^\top$, we have:

$\lambda_i \mathbf{u}_i = \sigma_i (A \mathbf{v}_i)$

Multiply both sides of the equation $A^\top \mathbf{u}_i = \sigma_i \mathbf{v}_i$ by $A^\top$ from the left:

$A^\top (A^\top \mathbf{u}_i) = A^\top (\sigma_i \mathbf{v}_i)$


$(A^\top A) \mathbf{v}_i = \sigma_i (A^\top \mathbf{u}_i)$

Because $A^\top \mathbf{u}_i = \sigma_i \mathbf{v}_i$:

$(A^\top A) \mathbf{v}_i = \sigma_i (\sigma_i \mathbf{v}_i)$

Because $\sigma_i = \sqrt{\lambda_i}$:

$(A^\top A) \mathbf{v}_i = (\sqrt{\lambda_i})(\sqrt{\lambda_i}) \mathbf{v}_i$

This simplifies to:

$(A^\top A) \mathbf{v}_i = \lambda_i \mathbf{v}_i$

Therefore, $\lambda_i$ and $\mathbf{v}_i$ are an eigenvalue and eigenvector of $A^\top A$.

Divide into 2 sides:

$\mathbf x \in N(AA^\top) \Rightarrow AA^\top \mathbf x=0 \Rightarrow (A^\top \mathbf x)^\top A^\top \mathbf x=0 \Rightarrow ||A^\top x||^2=0 \Rightarrow A^\top \mathbf x=0 \Rightarrow x \in N(A^\top)$

Therefore, $A A^\top$ has the same rank as $A$

4. 

Rewrite the given equation $A^\top \mathbf u_i = \sigma_i \mathbf v_i$ in matrix form as:

$A^\top U = V \Sigma$


Transpose both sides of the equation:

$(A^\top U)^\top = (V \Sigma)^\top$

Because $(AB)^\top = B^\top A^\top$, we have:

$U^\top A = \Sigma^\top V^\top$

Because $\Sigma$ is a diagonal matrix, so $\Sigma^\top = \Sigma$. Therefore:

$U^\top A = \Sigma V^\top$

Or equivalently:

$A = U \Sigma V^\top$

Here, $A$ is an $m \times n$ matrix, $U$ is an $m \times m$ matrix, $\Sigma$ is an $m \times n$ diagonal matrix, and $V$ is an $n \times n$ matrix.

5.

Start with $A^\top A$:

$A^\top A = (U \Sigma V^\top)^\top (U \Sigma V^\top)$

Using the property that $(AB)^\top = B^\top A^\top$, we get:

$A^\top A = (V \Sigma^\top U^\top) (U \Sigma V^\top)$

$A^\top A = V (\Sigma^\top U^\top U) \Sigma V^\top$

Since $U$ is an orthogonal matrix, $U^\top U = I_m$, where $I_m$ is the $m \times m$ identity matrix. Thus, we have:

$A^\top A = V (\Sigma^\top \Sigma) V^\top$

Since $\Sigma$ is a diagonal matrix, $\Sigma^\top \Sigma$ is also a diagonal matrix. The matrix $A^\top A$ is diagonalized by the product $V (\Sigma^\top \Sigma) V^\top$

Start next with $A A^\top$:

$A A^\top = (U \Sigma V^\top) (U \Sigma V^\top)^\top$

Using the property that $(AB)^\top = B^\top A^\top$, we get:

$A A^\top = (U \Sigma V^\top) (V \Sigma^\top U^\top)$


$A A^\top = U (\Sigma V^\top V) \Sigma^\top U^\top$

Since $V$ is an orthogonal matrix, $V^\top V = I_n$, where $I_n$ is the $n \times n$ identity matrix. Thus, we have:

$A A^\top = U (\Sigma \Sigma^\top) U^\top$

Since $\Sigma$ is a diagonal matrix, $\Sigma \Sigma^\top$ is also a diagonal matrix. Therefore, the matrix $A A^\top$ is diagonalized by the product $U (\Sigma \Sigma^\top) U^\top$

## 5.2 
Demean the data by subtracting each row by the mean. Compute the covariance matrix and its eigenvalues, and plot them in descending order. Describe your observations. Around how many values do we see the curve flattening, and how small are the eigenvalues at that point relative to the first one?

```python
# Use a smaller subset of the dataset
subset_size = 2500
X = X[np.random.choice(10000, subset_size, replace=False)].astype('float64')


# 5.2
# Demean the data
demeaned_data = X - mean

# Compute the covariance matrix
cov_matrix = np.cov(demeaned_data.T)

# Calculate the eigenvalues
eigenvalues, _ = np.linalg.eig(cov_matrix)

# Sort the eigenvalues in descending order
sorted_eigenvalues = np.sort(eigenvalues)[::-1]

# Plot the sorted eigenvalues
plt.plot(sorted_eigenvalues,'.')
plt.xlabel('Index')
plt.ylabel('Eigenvalue')
plt.show()
```
