Attempt to retrieve data and prevent rereading data

```python
# Attempt to retrieve the static reference data
%store -r static_reference_data
flag_to_stop_1 = False
flag_to_stop_2 = False
try:
    static_data = static_reference_data
    print("Static reference data has already been stored.")
    flag_to_stop_1 = True
except:
    print("Continue on to read the static data.")

# Attempt to retrieve the restaurant sales data
%store -r restaurant_sales_data
try:
    restaurant_data = restaurant_sales_data
    print("Restaurant sales data hs already been stored.")
    flag_to_stop_2 = True
except:
    print("Continue on to read the sales data.")

if flag_to_stop_1 and flag_to_stop_2:
    raise Exception("Some data has already been read and written!")
```

Read static reference data

```python
static_data = {}
for batch_index in [1,2]:
    static_data_filenames = os.listdir(f"palate_data_excel/batch_{batch_index}")
    for filename in tqdm(static_data_filenames, desc='Inner Loop', leave=False):
        if ".xlsx" in filename:
            df = pd.read_excel(f"palate_data_excel/batch_{batch_index}/{filename}")
            base_name = re.sub(r'\.xlsx$', '', filename)
            static_data[f"{base_name}_{batch_index}"] = df
```

Verify the location IDs in the static reference data match the location IDs for the sales data for each restaurant

```python
## Static Data
# Identify the list of location ids
locations = static_data['locations_1']['location_id'].value_counts().index.tolist()
# add on the second batch
locations.extend(static_data['locations_2']['location_id'].value_counts().index.tolist())

locations.sort()

## Restaurant Sales Data
# Concatenate the filenames
location_filenames = os.listdir("palate_data_excel/batch_1/orders_item_level") + os.listdir("palate_data_excel/batch_2/orders_item_level")
# Turn into ids
location_data = pd.Series(location_filenames).str.replace('.xlsx', '')

location_id_matching = pd.merge(pd.DataFrame(locations, columns=['location_id']), pd.DataFrame(location_data, columns=['location_data']), left_on='location_id', right_on='location_data', how='outer')
```

Write static reference data to parquet files

```python
# Write to parquet
for base_name, df in tqdm(static_data.items()):
    for column in df.columns:
        # Convert columns to string type for parquet files
        df[column] = df[column].astype(str)
    df.to_parquet(base_name + '.parquet')
```

Read sales data for each restaurant

```python
restaurant_data = {}
partial_data_filenames = []

# We will deal with this data file separately because it is too large for GitHub
large_data_location_id = "2HRX9P6HKXA8V"

# Read restaurant dataframes
for batch_index in [1,2]:
    location_filenames = os.listdir(f"palate_data_excel/batch_{batch_index}/orders_item_level")
    for location_filename in tqdm(location_filenames):
        location_id = re.sub(r'(_part[12])?\.xlsx$', '', location_filename)

            # Treat the large data file separately
            if "part" not in location_filename or location_id != large_data_location_id:
                df = pd.read_excel(f"palate_data_excel/batch_{batch_index}/orders_item_level/" + location_filename)
                restaurant_data[location_id] = df

            # If it is a partial dataframe, then select the first part
            elif "part1" in location_filename:
                partial_data_filenames.append(location_id)

# Read dataframes with a part1 and part2 then merge (apart from the large data file)
for location_id in partial_data_filenames:

        # Add both parts to a list to concat
        to_concat = []
        for part in [1,2]:
            to_concat.append(pd.read_excel(f"palate_data_excel/batch_1/orders_item_level/{location_id}_part{part}.xlsx"))
        restaurant_data[location_id] = pd.concat(to_concat)
```

Format data types

```python
for loc_id, df in tqdm(restaurant_data.items()):

    # Parquet files take string data types but not Pandas objects
    for column in df.columns:
        df[column] = df[column].astype(str)
        
    # Remove the alphabetical timezone identifier (there is a numerical one already)
    df['created_at'] = df['created_at'].str.replace(r'\s[A-Z]{3}', '', regex=True)
    df['created_at'] = pd.to_datetime(df['created_at'], utc=True)

    # Numerical data types
    df['item_price'] = df['item_price'].astype(int)
    df['item_quantity'] = df['item_quantity'].astype('float')
```

Write sales data to parquet files

```python
batch2_location_id_strings = "".join(os.listdir(f"palate_data_excel/batch_2/orders_item_level"))

# loop through and write to files
for loc_id, df in tqdm(restaurant_data.items()):

    # Keep the two batches separate for specificity's sake 
    if loc_id in batch2_location_id_strings:
        df.to_parquet(f"palate_data_parquet/batch_1/orders_item_level/{loc_id}.parquet")
    else:
        df.to_parquet(f"palate_data_parquet/batch_2/orders_item_level/{loc_id}.parquet")
```
