# Processing Data and Converting Excel to Parquet

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly
import scipy as sp
import statsmodels as sm
import sklearn as sk
import os
import re
import time
from tqdm import tqdm
```

Attempt to retrieve data and prevent rereading data

```python
%store -r static_reference_data
flag_to_stop_1 = False
flag_to_stop_2 = False
try:
    static_data = static_reference_data
    print("static reference data already stored")
    flag_to_stop_1 = True
except:
    print("continue on to read the static data")

%store -r restaurant_sales_data
try:
    restaurant_data = restaurant_sales_data
    print("restaurant sales data already stored")
    flag_to_stop_2 = True
except:
    print("continue on to read the sales data")

if flag_to_stop_1 and flag_to_stop_2:
    raise Exception("Data already read and written!")
```

Read static reference data

```python
static_data = {}
for batch_index in tqdm([1,2], desc='Outer Loop'):
    static_data_filenames = os.listdir(f"palate_data_excel/batch_{batch_index}")
    for filename in tqdm(static_data_filenames, desc='Inner Loop', leave=False):
        if ".xlsx" in filename:
            df = pd.read_excel(f"palate_data_excel/batch_{batch_index}/{filename}")
            base_name = re.sub(r'\.xlsx$', '', filename)
            static_data[f"{base_name}_{batch_index}"] = df
```

Verify there is data for each location id

```python
locations = static_data['locations_1']['location_id'].value_counts().index.tolist()
locations.extend(static_data['locations_2']['location_id'].value_counts().index.tolist())
locations.sort()
location_filenames = os.listdir("palate_data_excel/batch_1/orders_item_level") + os.listdir("palate_data_excel/batch_2/orders_item_level")
location_data = pd.Series(location_filenames).str.replace('.xlsx', '')
pd.merge(pd.DataFrame(locations, columns=['location_id']), pd.DataFrame(location_data, columns=['location_data']), left_on='location_id', right_on='location_data', how='outer')
```

Write static reference data to parquet files

```python
# prevent overwriting
%store -r static_reference_data
try:
    static_data = static_reference_data
    print("static reference data already stored")
except:
    print("continue on to read the static data")

for base_name, df in tqdm(static_data.items()):
    for column in df.columns:
        df[column] = df[column].astype(str)
    df.to_parquet(base_name + '.parquet')
```

Read sales data for each restaurant

```python
restaurant_data = {}
part1_files = {}
# Read restaurant dataframes
for batch_index in [1,2]:
    location_filenames = os.listdir(f"palate_data_excel/batch_{batch_index}/orders_item_level")
    for location_filename in tqdm(location_filenames):
        location_id = re.sub(r'(_part[12])?\.xlsx$', '', location_filename)
        if "part1" in location_filename:
            part1_files[location_id] = location_filename
        else:
            df = pd.read_excel(f"palate_data_excel/batch_{batch_index}/orders_item_level/" + location_filename)
            restaurant_data[location_id] = df
    
# Read dataframes with a part1 and part2 then merge
for location_id, part1_filename in part1_files.items():
    df1 = pd.read_excel(f"palate_data_excel/batch_1/orders_item_level/" + part1_filename)
    part2_filename = f"{location_id}_part2.xlsx"
    df2 = pd.read_excel(f"palate_data_excel/batch_1/orders_item_level/" + part2_filename)
    restaurant_data[location_id] = pd.concat([df1, df2])
```

Keep largest file split

```python
largedf_part1 = pd.read_excel(f"palate_data_excel/batch_1/orders_item_level/2HRX9P6HKXA8V_part1.xlsx")
largedf_part2 = pd.read_excel(f"palate_data_excel/batch_1/orders_item_level/2HRX9P6HKXA8V_part2.xlsx")
for column in largedf_part1.columns:
    largedf_part1[column] = largedf_part1[column].astype(str)
    largedf_part2[column] = largedf_part2[column].astype(str)
largedf_part1['created_at'] = largedf_part1['created_at'].str.replace(r'\s[A-Z]{3}', '', regex=True)
largedf_part1['created_at'] = pd.to_datetime(largedf_part1['created_at'], utc=True)
largedf_part2['created_at'] = largedf_part2['created_at'].str.replace(r'\s[A-Z]{3}', '', regex=True)
largedf_part2['created_at'] = pd.to_datetime(largedf_part2['created_at'], utc=True)
largedf_part1['item_price'] = largedf_part1['item_price'].astype(int)
largedf_part2['item_price'] = largedf_part2['item_price'].astype(int)
largedf_part1['item_quantity'] = largedf_part1['item_quantity'].astype('float')
largedf_part2['item_quantity'] = largedf_part2['item_quantity'].astype('float')
largedf_part1.to_parquet("palate_data_parquet/batch_1/orders_item_level/2HRX9P6HKXA8V_part1.parquet")
largedf_part2.to_parquet("palate_data_parquet/batch_1/orders_item_level/2HRX9P6HKXA8V_part2.parquet")
```

Write sales data to parquet files

```python
# prevent overwriting
%store -r restaurant_sales_data
try:
    restaurant_data = restaurant_sales_data
    print("data already stored")
except:
    print("continue on to read the data")

# loop through and write to files
batch2string = "".join(os.listdir(f"palate_data_excel/batch_2/orders_item_level"))
for location_id, df in tqdm(restaurant_data.items()):
    if location_id != "2HRX9P6HKXA8V":
        for column in df.columns:
            df[column] = df[column].astype(str)
        df['created_at'] = df['created_at'].str.replace(r'\s[A-Z]{3}', '', regex=True)
        df['created_at'] = pd.to_datetime(df['created_at'], utc=True)
        df['item_price'] = df['item_price'].astype(int)
        df['item_quantity'] = df['item_quantity'].astype('float')
        if location_id not in batch2string:
            df.to_parquet(f"palate_data_parquet/batch_1/orders_item_level/{location_id}.parquet")
        else:
            df.to_parquet(f"palate_data_parquet/batch_2/orders_item_level/{location_id}.parquet")
```
