## Simple linear regression

**Setting:** We have 250 observations ($n=250$). An input $X \in \mathbb{R}$ follows a standard Gaussian distribution. An output $Y$ is generated as follows.
\begin{align*}
    Y = \beta_0 + \beta_1 X +\epsilon,
\end{align*}
where $\epsilon$ indicates a random error with a standard Gaussian distribution and we assume $(\beta_0, \beta_1) = (1.5, 2.5)$.

1) Generate the data 

```python
import numpy as np
```

```python
n = 250
beta0 = 1.5
beta1 = 2.5
x = np.random.normal(0, 1, n)
y = beta0 + x*beta1 + np.random.normal(0, 1, n)
xbar = np.sum(x)/n
ybar = np.sum(y)/n
sxx = np.sum((x - xbar)**2)
```

2) Implement the OLS estimator.

\begin{align*}
    (\mathbb{X}^T \mathbb{X})^{-1} (\mathbb{X}^T \mathbb{Y})
    &= \left( \begin{bmatrix}
    n & \sum_{i=1} ^n x_{i1}\\
    \sum_{i=1} ^n x_{i1} & \sum_{i=1} ^n x_{i1} ^2 
    \end{bmatrix} \right)^{-1}
    \begin{bmatrix}
    \sum_{i=1} ^n y_i \\
    \sum_{i=1} ^n x_{i1} y_i 
    \end{bmatrix} \\
    &= \frac{1}{n(\sum_{i=1} ^n x_{i1} ^2) - (\sum_{i=1} ^n x_{i1})^2}  \begin{bmatrix}
    \sum_{i=1} ^n x_{i1} ^2 & -\sum_{i=1} ^n x_{i1}\\
    -\sum_{i=1} ^n x_{i1} & n 
    \end{bmatrix}
    \begin{bmatrix}
    \sum_{i=1} ^n y_i \\
    \sum_{i=1} ^n x_{i1} y_i 
    \end{bmatrix}.
\end{align*}

```python
beta1hat = (np.sum(x*y)-(1/n)*np.sum(x)*np.sum(y))/sxx
beta0hat = ybar - beta1hat*xbar

print(beta0hat)
print(beta1hat)
```

## Multivariate linear regression

**Setting:** We have 250 observations ($n=250$). An input $X=(X_1, \dots, X_5)^T \in \mathbb{R}^5$ is 5-dimensional random vector and follows multivariate Gaussian distribution with mean zero and the 5-dimensional identity matrix. An output $Y$ is generated as follows.
\begin{align*}
    Y = \beta_0 + \sum_{j=1} ^5 \beta_j X_j +\epsilon,
\end{align*}
where $\epsilon$ indicates a random error with a standard Gaussian distribution and we assume $\beta_0 =1.5$ and $\beta_j = 3 - 0.5 \times j$ for $j \in \{1, \dots, 5\}$.


1) Generate the data 

```python
mean = [0,0,0,0,0]
cov = np.eye(5)
X = np.random.multivariate_normal(mean, cov, 250)
e = np.random.normal(0, 1, 250)
X = np.hstack((np.ones(shape=(n,1)),X))
beta0 = 1.5
betaj = [3 - .5*j for j in range(1,6)]
betaj.insert(0,beta0)
beta = np.array(betaj)
y = X.dot(beta) + e
```

2) Implement the OLS estimator. Hint: `np.linalg.inv`

```python
gram_matrix = np.matmul(X.T, X)
first = np.linalg.inv(gram_matrix)
second = X.T.dot(y)
betahat = np.array([1,1,1,1,1,1])
betahat = first.dot(second)
beta = beta.astype(np.double).reshape((1,6))
betahat = betahat.astype(np.double).reshape((1,6))
np.concatenate([beta, betahat])
```

3) Compute the L2 distance between the estimator you obtained from problem 1) and the true value.


```python
np.sqrt(np.sum((beta - betahat)**2))
```
